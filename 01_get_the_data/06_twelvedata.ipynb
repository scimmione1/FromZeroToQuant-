```python
from dotenv import load_dotenv
import os
from twelvedata import TDClient

load_dotenv()
api_key = os.getenv("TWELVEDATA_API_KEY")

if not api_key:
    raise ValueError("Missing TWELVEDATA_API_KEY in the .env file")

td = TDClient(apikey=api_key)

ts = td.time_series(
    symbol="NVDA",
    interval="1day",
    outputsize=365,
    timezone="America/New_York",
)

df = ts.as_pandas()
df.head()
```
